```python
import pandas as pd
from pathlib import Path

root = Path("/content/official_final_test_results")

runs = {
    "ResNet50": "20260901_065938_finaltest_resnet50_seed42_2e003a",
    "MaxViT": "20260901_070357_finaltest_maxvit_seed42_a073c3",
    "EfficientNetB0": "20260901_070518_finaltest_efficientnetb0_seed42_6a24d8",
    "AA-EvidentNet": "20260901_070854_finaltest_aa_evidentnet_seed42_5ad97f",
}

dfs = {}

for model, run_id in runs.items():
    path = root / run_id / "predictions.csv"
    dfs[model] = pd.read_csv(path)

    print(
        model,
        "rows =", len(dfs[model]),
        "| unique sample IDs =", dfs[model]["sample_id"].nunique()
    )

reference = dfs["AA-EvidentNet"]

print("\nPAIRING CHECK")

for model, df in dfs.items():
    same_ids = df["sample_id"].tolist() == reference["sample_id"].tolist()
    same_true_idx = (
        df["true_class_index"].tolist()
        == reference["true_class_index"].tolist()
    )
    same_true_names = (
        df["true_class_name"].tolist()
        == reference["true_class_name"].tolist()
    )

    print(
        model,
        "| same IDs:", same_ids,
        "| same true indices:", same_true_idx,
        "| same true names:", same_true_names
    )
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, balanced_accuracy_score, f1_score
from statsmodels.stats.contingency_tables import mcnemar
from statsmodels.stats.multitest import multipletests

# --------------------------------------------------
# 1. Bootstrap 95% confidence intervals
# --------------------------------------------------

N_BOOTSTRAP = 10000
SEED = 42
rng = np.random.default_rng(SEED)

def metric_values(y_true, y_pred):
    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Balanced Accuracy": balanced_accuracy_score(y_true, y_pred),
        "Macro-F1": f1_score(y_true, y_pred, average="macro", zero_division=0),
    }

bootstrap_rows = []

for model, df in dfs.items():
    y_true = df["true_class_index"].to_numpy()
    y_pred = df["predicted_class_index"].to_numpy()
    n = len(df)

    observed = metric_values(y_true, y_pred)

    boot = {
        "Accuracy": [],
        "Balanced Accuracy": [],
        "Macro-F1": [],
    }

    for _ in range(N_BOOTSTRAP):
        idx = rng.integers(0, n, size=n)

        values = metric_values(
            y_true[idx],
            y_pred[idx]
        )

        for metric in boot:
            boot[metric].append(values[metric])

    for metric in observed:
        lower, upper = np.percentile(
            boot[metric],
            [2.5, 97.5]
        )

        bootstrap_rows.append({
            "Model": model,
            "Metric": metric,
            "Estimate": observed[metric],
            "CI Lower": lower,
            "CI Upper": upper,
        })

bootstrap_df = pd.DataFrame(bootstrap_rows)

# --------------------------------------------------
# 2. McNemar: AA-EvidentNet vs each baseline
# --------------------------------------------------

aa = dfs["AA-EvidentNet"]
aa_correct = aa["correct"].astype(bool).to_numpy()

mcnemar_rows = []

for baseline in ["ResNet50", "MaxViT", "EfficientNetB0"]:
    baseline_correct = dfs[baseline]["correct"].astype(bool).to_numpy()

    # Discordant cases:
    # baseline wrong, AA correct
    aa_wins = int(np.sum(~baseline_correct & aa_correct))

    # baseline correct, AA wrong
    baseline_wins = int(np.sum(baseline_correct & ~aa_correct))

    table = [
        [
            int(np.sum(~baseline_correct & ~aa_correct)),
            aa_wins
        ],
        [
            baseline_wins,
            int(np.sum(baseline_correct & aa_correct))
        ]
    ]

    # Exact McNemar is appropriate and avoids asymptotic issues
    result = mcnemar(table, exact=True)

    mcnemar_rows.append({
        "Comparison": f"AA-EvidentNet vs {baseline}",
        "AA correct / baseline wrong": aa_wins,
        "Baseline correct / AA wrong": baseline_wins,
        "Raw p-value": result.pvalue,
    })

mcnemar_df = pd.DataFrame(mcnemar_rows)

# --------------------------------------------------
# 3. Holm multiple-comparison correction
# --------------------------------------------------

reject, corrected_p, _, _ = multipletests(
    mcnemar_df["Raw p-value"],
    alpha=0.05,
    method="holm"
)

mcnemar_df["Holm-adjusted p-value"] = corrected_p
mcnemar_df["Significant after Holm (alpha=0.05)"] = reject

# --------------------------------------------------
# Display results
# --------------------------------------------------

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

print("\nBOOTSTRAP 95% CONFIDENCE INTERVALS")
print("=" * 80)

print(
    bootstrap_df.to_string(
        index=False,
        formatters={
            "Estimate": "{:.4f}".format,
            "CI Lower": "{:.4f}".format,
            "CI Upper": "{:.4f}".format,
        }
    )
)

print("\n\nMCNEMAR TESTS")
print("=" * 80)

print(
    mcnemar_df.to_string(
        index=False,
        formatters={
            "Raw p-value": "{:.6f}".format,
            "Holm-adjusted p-value": "{:.6f}".format,
        }
    )
)

# --------------------------------------------------
# Save for the paper
# --------------------------------------------------

bootstrap_df.to_csv(
    "/content/bootstrap_95ci.csv",
    index=False
)

mcnemar_df.to_csv(
    "/content/mcnemar_comparisons.csv",
    index=False
)

print("\nSaved:")
print("/content/bootstrap_95ci.csv")
print("/content/mcnemar_comparisons.csv")
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
)
from statsmodels.stats.contingency_tables import mcnemar
from statsmodels.stats.multitest import multipletests

# ============================================================
# 1. Bootstrap 95% confidence intervals
# ============================================================

N_BOOTSTRAP = 10_000
SEED = 42

def calculate_metrics(y_true, y_pred):
    return {
        "Accuracy": accuracy_score(y_true, y_pred),
        "Balanced Accuracy": balanced_accuracy_score(y_true, y_pred),
        "Macro-F1": f1_score(
            y_true,
            y_pred,
            average="macro",
            zero_division=0,
        ),
    }

bootstrap_rows = []

for model, df in dfs.items():

    y_true = df["true_class_index"].to_numpy()
    y_pred = df["predicted_class_index"].to_numpy()
    n = len(df)

    observed = calculate_metrics(y_true, y_pred)

    # Separate deterministic RNG for each model
    rng = np.random.default_rng(SEED)

    boot_values = {
        "Accuracy": np.empty(N_BOOTSTRAP),
        "Balanced Accuracy": np.empty(N_BOOTSTRAP),
        "Macro-F1": np.empty(N_BOOTSTRAP),
    }

    for b in range(N_BOOTSTRAP):
        idx = rng.integers(0, n, size=n)

        metrics = calculate_metrics(
            y_true[idx],
            y_pred[idx],
        )

        for metric in boot_values:
            boot_values[metric][b] = metrics[metric]

    for metric, estimate in observed.items():

        lower, upper = np.percentile(
            boot_values[metric],
            [2.5, 97.5],
        )

        bootstrap_rows.append({
            "Model": model,
            "Metric": metric,
            "Estimate": estimate,
            "CI Lower": lower,
            "CI Upper": upper,
        })

bootstrap_df = pd.DataFrame(bootstrap_rows)

# ============================================================
# 2. Exact McNemar tests
#    AA-EvidentNet vs each baseline
# ============================================================

aa_correct = dfs["AA-EvidentNet"]["correct"].astype(bool).to_numpy()

mcnemar_rows = []

for baseline in [
    "ResNet50",
    "MaxViT",
    "EfficientNetB0",
]:

    baseline_correct = (
        dfs[baseline]["correct"]
        .astype(bool)
        .to_numpy()
    )

    # Discordant samples
    aa_wins = int(
        np.sum(aa_correct & ~baseline_correct)
    )

    baseline_wins = int(
        np.sum(~aa_correct & baseline_correct)
    )

    both_wrong = int(
        np.sum(~aa_correct & ~baseline_correct)
    )

    both_correct = int(
        np.sum(aa_correct & baseline_correct)
    )

    # Rows = baseline wrong/correct
    # Cols = AA wrong/correct
    table = [
        [both_wrong, aa_wins],
        [baseline_wins, both_correct],
    ]

    result = mcnemar(
        table,
        exact=True,
    )

    mcnemar_rows.append({
        "Comparison": f"AA-EvidentNet vs {baseline}",
        "AA correct / baseline wrong": aa_wins,
        "Baseline correct / AA wrong": baseline_wins,
        "Raw p-value": result.pvalue,
    })

mcnemar_df = pd.DataFrame(mcnemar_rows)

# ============================================================
# 3. Holm correction for the 3 comparisons
# ============================================================

reject, adjusted_p, _, _ = multipletests(
    mcnemar_df["Raw p-value"].to_numpy(),
    alpha=0.05,
    method="holm",
)

mcnemar_df["Holm-adjusted p-value"] = adjusted_p
mcnemar_df["Significant after Holm"] = reject

# ============================================================
# 4. Display
# ============================================================

print("\nBOOTSTRAP 95% CONFIDENCE INTERVALS")
print("=" * 90)

print(
    bootstrap_df.to_string(
        index=False,
        formatters={
            "Estimate": "{:.4f}".format,
            "CI Lower": "{:.4f}".format,
            "CI Upper": "{:.4f}".format,
        },
    )
)

print("\n\nEXACT MCNEMAR TESTS")
print("=" * 90)

print(
    mcnemar_df.to_string(
        index=False,
        formatters={
            "Raw p-value": "{:.6f}".format,
            "Holm-adjusted p-value": "{:.6f}".format,
        },
    )
)

# ============================================================
# 5. Save results
# ============================================================

bootstrap_df.to_csv(
    "/content/bootstrap_95ci.csv",
    index=False,
)

mcnemar_df.to_csv(
    "/content/mcnemar_comparisons.csv",
    index=False,
)

print("\nSaved:")
print("/content/bootstrap_95ci.csv")
print("/content/mcnemar_comparisons.csv")
```

```python
import pandas as pd

hard_pairs = [
    ("Healthy", "Glaucoma"),
    ("Disc Edema", "Glaucoma"),
    ("Diabetic Retinopathy", "Central Serous Chorioretinopathy"),
]

rows = []

for model, df in dfs.items():

    for class_a, class_b in hard_pairs:

        # True A predicted as B
        a_to_b = (
            (df["true_class_name"] == class_a) &
            (df["predicted_class_name"] == class_b)
        ).sum()

        # True B predicted as A
        b_to_a = (
            (df["true_class_name"] == class_b) &
            (df["predicted_class_name"] == class_a)
        ).sum()

        support_a = (df["true_class_name"] == class_a).sum()
        support_b = (df["true_class_name"] == class_b).sum()

        total_pair_confusions = a_to_b + b_to_a

        rows.append({
            "Model": model,
            "Hard Pair": f"{class_a} ↔ {class_b}",
            f"{class_a} → {class_b}": int(a_to_b),
            f"{class_b} → {class_a}": int(b_to_a),
            "Total Pair Confusions": int(total_pair_confusions),
            "Pair Support": int(support_a + support_b),
            "Pair Confusion Rate": total_pair_confusions / (support_a + support_b),
        })

hard_pair_df = pd.DataFrame(rows)

print(
    hard_pair_df.to_string(
        index=False,
        formatters={
            "Pair Confusion Rate": "{:.4f}".format
        }
    )
)

hard_pair_df.to_csv(
    "/content/hard_pair_analysis.csv",
    index=False
)

print("\nSaved:")
print("/content/hard_pair_analysis.csv")
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score, average_precision_score

N_BINS = 15
N_CLASSES = 10


def expected_calibration_error(y_true, probs, n_bins=15):
    confidence = probs.max(axis=1)
    prediction = probs.argmax(axis=1)
    correct = (prediction == y_true).astype(float)

    bins = np.linspace(0.0, 1.0, n_bins + 1)
    ece = 0.0

    for i in range(n_bins):
        if i == 0:
            mask = (confidence >= bins[i]) & (confidence <= bins[i + 1])
        else:
            mask = (confidence > bins[i]) & (confidence <= bins[i + 1])

        if not np.any(mask):
            continue

        bin_accuracy = correct[mask].mean()
        bin_confidence = confidence[mask].mean()

        ece += (
            mask.mean()
            * abs(bin_accuracy - bin_confidence)
        )

    return ece


def multiclass_brier(y_true, probs, n_classes=10):
    one_hot = np.eye(n_classes)[y_true]

    # Mean squared distance between probability vector
    # and one-hot target.
    return np.mean(
        np.sum((probs - one_hot) ** 2, axis=1)
    )


rows = []

for model, df in dfs.items():

    y_true = df["true_class_index"].to_numpy(dtype=int)
    error = (~df["correct"].astype(bool)).astype(int).to_numpy()

    prob_cols = [f"prob_{i}" for i in range(N_CLASSES)]
    probs = df[prob_cols].to_numpy(dtype=float)

    ece = expected_calibration_error(
        y_true,
        probs,
        n_bins=N_BINS,
    )

    brier = multiclass_brier(
        y_true,
        probs,
        n_classes=N_CLASSES,
    )

    # Higher score should mean "more likely to be wrong".
    softmax_error_score = 1.0 - probs.max(axis=1)

    softmax_auroc = roc_auc_score(
        error,
        softmax_error_score,
    )

    softmax_auprc = average_precision_score(
        error,
        softmax_error_score,
    )

    rows.append({
        "Model": model,
        "Probability Type": "Standard probabilities",
        "ECE": ece,
        "Brier Score": brier,
        "Error AUROC": softmax_auroc,
        "Error AUPRC": softmax_auprc,
    })

    # Additional EDL evaluation for AA-EvidentNet
    if model == "AA-EvidentNet":

        edl_cols = [
            f"evidential_prob_{i}"
            for i in range(N_CLASSES)
        ]

        edl_probs = df[edl_cols].to_numpy(dtype=float)
        uncertainty = df["uncertainty"].to_numpy(dtype=float)

        edl_ece = expected_calibration_error(
            y_true,
            edl_probs,
            n_bins=N_BINS,
        )

        edl_brier = multiclass_brier(
            y_true,
            edl_probs,
            n_classes=N_CLASSES,
        )

        edl_auroc = roc_auc_score(
            error,
            uncertainty,
        )

        edl_auprc = average_precision_score(
            error,
            uncertainty,
        )

        rows.append({
            "Model": model,
            "Probability Type": "Evidential / Dirichlet",
            "ECE": edl_ece,
            "Brier Score": edl_brier,
            "Error AUROC": edl_auroc,
            "Error AUPRC": edl_auprc,
        })


calibration_df = pd.DataFrame(rows)

print("\nCALIBRATION + ERROR DETECTION")
print("=" * 105)

print(
    calibration_df.to_string(
        index=False,
        formatters={
            "ECE": "{:.4f}".format,
            "Brier Score": "{:.4f}".format,
            "Error AUROC": "{:.4f}".format,
            "Error AUPRC": "{:.4f}".format,
        }
    )
)

calibration_df.to_csv(
    "/content/calibration_error_detection.csv",
    index=False,
)

print("\nSaved:")
print("/content/calibration_error_detection.csv")
```

```python
import numpy as np
import pandas as pd

aa = dfs["AA-EvidentNet"].copy()

# Lower uncertainty = more certain.
aa = aa.sort_values(
    "uncertainty",
    ascending=True
).reset_index(drop=True)

coverages = [
    1.00,
    0.95,
    0.90,
    0.85,
    0.80,
    0.75,
    0.70,
    0.60,
    0.50,
]

rows = []
n_total = len(aa)

for coverage in coverages:

    n_keep = int(np.ceil(n_total * coverage))

    kept = aa.iloc[:n_keep]

    accuracy = kept["correct"].astype(bool).mean()
    risk = 1.0 - accuracy

    referred = n_total - n_keep

    rows.append({
        "Coverage": coverage,
        "Samples Kept": n_keep,
        "Samples Referred": referred,
        "Selective Accuracy": accuracy,
        "Selective Risk": risk,
        "Max Uncertainty Kept": kept["uncertainty"].max(),
    })

selective_df = pd.DataFrame(rows)

print("\nAA-EVIDENTNET SELECTIVE REFERRAL")
print("=" * 90)

print(
    selective_df.to_string(
        index=False,
        formatters={
            "Coverage": "{:.0%}".format,
            "Selective Accuracy": "{:.4f}".format,
            "Selective Risk": "{:.4f}".format,
            "Max Uncertainty Kept": "{:.6f}".format,
        }
    )
)

selective_df.to_csv(
    "/content/selective_referral.csv",
    index=False,
)

print("\nSaved:")
print("/content/selective_referral.csv")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

aa = dfs["AA-EvidentNet"]
y_true = aa["true_class_index"].to_numpy(dtype=int)

standard_probs = aa[
    [f"prob_{i}" for i in range(10)]
].to_numpy(dtype=float)

edl_probs = aa[
    [f"evidential_prob_{i}" for i in range(10)]
].to_numpy(dtype=float)


def adaptive_reliability(y_true, probs, n_bins=10):
    confidence = probs.max(axis=1)
    prediction = probs.argmax(axis=1)
    correct = (prediction == y_true).astype(float)

    # Sort from lowest to highest confidence
    order = np.argsort(confidence)

    # Split into approximately equal-size groups
    groups = np.array_split(order, n_bins)

    mean_confidence = []
    observed_accuracy = []
    counts = []

    for group in groups:
        mean_confidence.append(confidence[group].mean())
        observed_accuracy.append(correct[group].mean())
        counts.append(len(group))

    return (
        np.array(mean_confidence),
        np.array(observed_accuracy),
        np.array(counts),
    )


std_conf, std_acc, std_n = adaptive_reliability(
    y_true,
    standard_probs
)

edl_conf, edl_acc, edl_n = adaptive_reliability(
    y_true,
    edl_probs
)

plt.figure(figsize=(7, 7))

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Perfect calibration"
)

plt.plot(
    std_conf,
    std_acc,
    marker="o",
    linewidth=2,
    label="AA-EvidentNet (standard)"
)

plt.plot(
    edl_conf,
    edl_acc,
    marker="s",
    linewidth=2,
    label="AA-EvidentNet (evidential)"
)

plt.xlabel("Mean Predicted Confidence")
plt.ylabel("Observed Accuracy")
plt.title("AA-EvidentNet Adaptive Reliability Diagram")
plt.xlim(0, 1)
plt.ylim(0, 1)
plt.grid(alpha=0.25)
plt.legend()
plt.tight_layout()

fig_dir = Path("/content/paper_figures")
png = fig_dir / "aa_evidentnet_adaptive_reliability.png"
pdf = fig_dir / "aa_evidentnet_adaptive_reliability.pdf"

plt.savefig(png, dpi=300, bbox_inches="tight")
plt.savefig(pdf, bbox_inches="tight")
plt.show()

print("Standard samples per bin:", std_n.tolist())
print("Evidential samples per bin:", edl_n.tolist())

print("\nSaved:")
print(png)
print(pdf)
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import confusion_matrix

class_names = [
    "Central Serous Chorioretinopathy",
    "Diabetic Retinopathy",
    "Disc Edema",
    "Glaucoma",
    "Healthy",
    "Macular Scar",
    "Myopia",
    "Pterygium",
    "Retinal Detachment",
    "Retinitis Pigmentosa",
]

rows = []

for model, df in dfs.items():

    y_true = df["true_class_index"].to_numpy(dtype=int)
    y_pred = df["predicted_class_index"].to_numpy(dtype=int)

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=list(range(10))
    )

    total = cm.sum()

    for i, class_name in enumerate(class_names):

        tp = cm[i, i]
        fn = cm[i, :].sum() - tp
        fp = cm[:, i].sum() - tp
        tn = total - tp - fn - fp

        precision = tp / (tp + fp) if (tp + fp) > 0 else np.nan
        recall = tp / (tp + fn) if (tp + fn) > 0 else np.nan
        specificity = tn / (tn + fp) if (tn + fp) > 0 else np.nan

        f1 = (
            2 * precision * recall / (precision + recall)
            if precision + recall > 0
            else 0.0
        )

        support = tp + fn

        rows.append({
            "Model": model,
            "Class": class_name,
            "Support": int(support),
            "Precision": precision,
            "Recall": recall,
            "Specificity": specificity,
            "F1": f1,
        })

per_class_df = pd.DataFrame(rows)

print(
    per_class_df.to_string(
        index=False,
        formatters={
            "Precision": "{:.4f}".format,
            "Recall": "{:.4f}".format,
            "Specificity": "{:.4f}".format,
            "F1": "{:.4f}".format,
        }
    )
)

per_class_df.to_csv(
    "/content/per_class_metrics_all_models.csv",
    index=False
)

print("\nSaved:")
print("/content/per_class_metrics_all_models.csv")
```

```python
%cd /content/AA-EvidentNet

import torch
import torch.nn.functional as F

class GradCAM:
    def __init__(self, model, target_layer):
        self.model = model
        self.target_layer = target_layer

        self.activations = None
        self.gradients = None

        self.forward_handle = target_layer.register_forward_hook(
            self._save_activations
        )

        self.backward_handle = target_layer.register_full_backward_hook(
            self._save_gradients
        )

    def _save_activations(self, module, inputs, output):
        self.activations = output

    def _save_gradients(self, module, grad_input, grad_output):
        self.gradients = grad_output[0]

    def generate(self, image_tensor, class_idx=None):
        self.model.zero_grad(set_to_none=True)

        output = self.model(image_tensor, return_features=True)
        logits = output.logits

        if class_idx is None:
            class_idx = int(logits.argmax(dim=1).item())

        score = logits[:, class_idx].sum()
        score.backward()

        activations = self.activations
        gradients = self.gradients

        # Standard Grad-CAM channel weights.
        weights = gradients.mean(dim=(2, 3), keepdim=True)

        cam = (weights * activations).sum(dim=1, keepdim=True)
        cam = F.relu(cam)

        # Resize to input resolution.
        cam = F.interpolate(
            cam,
            size=image_tensor.shape[-2:],
            mode="bilinear",
            align_corners=False,
        )

        # Normalize to [0, 1].
        cam_min = cam.amin(dim=(2, 3), keepdim=True)
        cam_max = cam.amax(dim=(2, 3), keepdim=True)

        cam = (cam - cam_min) / (cam_max - cam_min + 1e-8)

        return cam.detach().cpu().squeeze().numpy(), class_idx, output

    def remove_hooks(self):
        self.forward_handle.remove()
        self.backward_handle.remove()


global_cam = GradCAM(
    model,
    model.global_backbone.stages[3]
)

local_cam = GradCAM(
    model,
    model.local_branch.conv_blocks[9]
)

print("Grad-CAM initialized.")
print("Global target: global_backbone.stages[3]")
print("Local target: local_branch.conv_blocks[9]")
```

```python
# Run a fresh forward/backward pass and inspect the raw Grad-CAM signals.

model.zero_grad(set_to_none=True)

output = model(x, return_features=True)
class_idx = int(output.logits.argmax(dim=1).item())

output.logits[:, class_idx].sum().backward()

for name, cam_obj in [
    ("GLOBAL", global_cam),
    ("LOCAL", local_cam),
]:
    acts = cam_obj.activations.detach()
    grads = cam_obj.gradients.detach()

    weights = grads.mean(dim=(2, 3), keepdim=True)
    raw_cam = (weights * acts).sum(dim=1)

    print(f"\n=== {name} ===")
    print("Activations shape:", tuple(acts.shape))
    print("Gradients shape:", tuple(grads.shape))

    print(
        "Gradient |mean|:",
        float(grads.abs().mean())
    )

    print(
        "Gradient |max|:",
        float(grads.abs().max())
    )

    print(
        "Raw CAM min:",
        float(raw_cam.min())
    )

    print(
        "Raw CAM max:",
        float(raw_cam.max())
    )

    print(
        "Positive raw CAM values:",
        int((raw_cam > 0).sum())
    )
```

```python
import pandas as pd

preds = pd.read_csv(
    "/content/AA-EvidentNet/results/raw_predictions/"
    "20260901_095430_finaltest_aa_evidentnet_seed42_86b8d2/"
    "predictions.csv"
)

# 1. Deterministic random correct prediction
correct_pool = preds[preds["correct"] == True]
random_correct = correct_pool.sample(n=1, random_state=42).iloc[0]

# 2. Healthy <-> Glaucoma hard-pair error
hg_error_pool = preds[
    (~preds["correct"]) &
    (
        (
            (preds["true_class_name"] == "Healthy") &
            (preds["predicted_class_name"] == "Glaucoma")
        )
        |
        (
            (preds["true_class_name"] == "Glaucoma") &
            (preds["predicted_class_name"] == "Healthy")
        )
    )
]

hard_pair_error = (
    hg_error_pool.sort_values("sample_id").iloc[0]
    if len(hg_error_pool) > 0
    else None
)

# 3. Highest-uncertainty correct prediction
highest_unc_correct = (
    correct_pool
    .sort_values(["uncertainty", "sample_id"], ascending=[False, True])
    .iloc[0]
)

# 4. Highest-uncertainty incorrect prediction
incorrect_pool = preds[preds["correct"] == False]

highest_unc_incorrect = (
    incorrect_pool
    .sort_values(["uncertainty", "sample_id"], ascending=[False, True])
    .iloc[0]
)

selected = [
    ("Random correct", random_correct),
    ("Healthy-Glaucoma error", hard_pair_error),
    ("Highest-uncertainty correct", highest_unc_correct),
    ("Highest-uncertainty incorrect", highest_unc_incorrect),
]

for category, row in selected:
    print("\n===", category, "===")

    if row is None:
        print("No eligible sample.")
        continue

    print("Sample ID:", row["sample_id"])
    print("True:", row["true_class_name"])
    print("Predicted:", row["predicted_class_name"])
    print("Correct:", row["correct"])
    print("Uncertainty:", row["uncertainty"])
    print("Image:", row["image_path"])
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix
from pathlib import Path

preds = pd.read_csv(
    "/content/AA-EvidentNet/results/raw_predictions/"
    "20260901_095430_finaltest_aa_evidentnet_seed42_86b8d2/"
    "predictions.csv"
)

class_names = [
    "CSC", "DR", "DE", "GL", "HE",
    "MS", "MY", "PT", "RD", "RP"
]

cm = confusion_matrix(
    preds["true_class_index"],
    preds["predicted_class_index"],
    labels=range(10)
)

# Row-normalized proportions
cm_norm = cm / cm.sum(axis=1, keepdims=True)

fig, ax = plt.subplots(figsize=(10, 8))

im = ax.imshow(
    cm_norm,
    cmap="Blues",
    vmin=0,
    vmax=1
)

for i in range(10):
    for j in range(10):

        count = cm[i, j]
        proportion = cm_norm[i, j]

        text_color = "white" if proportion > 0.5 else "black"

        ax.text(
            j,
            i,
            f"{count}\n({proportion:.2f})",
            ha="center",
            va="center",
            fontsize=7,
            color=text_color
        )

ax.set_xticks(range(10))
ax.set_yticks(range(10))

ax.set_xticklabels(class_names)
ax.set_yticklabels(class_names)

ax.set_xlabel("Predicted Class")
ax.set_ylabel("True Class")

ax.set_title(
    "AA-EvidentNet Confusion Matrix on the Held-Out Test Set"
)

fig.colorbar(
    im,
    ax=ax,
    label="Row-normalized proportion"
)

fig.tight_layout()

out_dir = Path("/content/paper_figures")

png = out_dir / "aa_evidentnet_confusion_matrix_final.png"
pdf = out_dir / "aa_evidentnet_confusion_matrix_final.pdf"

fig.savefig(png, dpi=300, bbox_inches="tight")
fig.savefig(pdf, bbox_inches="tight")

plt.show()

print("Test samples:", cm.sum())
print("Correct:", np.trace(cm))
print("Errors:", cm.sum() - np.trace(cm))

print("Created:", png)
print("Created:", pdf)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

preds = pd.read_csv(
    "/content/AA-EvidentNet/results/raw_predictions/"
    "20260901_095430_finaltest_aa_evidentnet_seed42_86b8d2/"
    "predictions.csv"
)

# Evidential confidence = maximum expected Dirichlet probability
evidential_cols = [
    f"evidential_prob_{i}" for i in range(10)
]

confidence = preds[evidential_cols].max(axis=1).to_numpy()
correct = preds["correct"].astype(float).to_numpy()

# Equal-frequency bins for visualization only
order = np.argsort(confidence)
bins = np.array_split(order, 10)

bin_confidence = []
bin_accuracy = []
bin_size = []

for idx in bins:
    bin_confidence.append(confidence[idx].mean())
    bin_accuracy.append(correct[idx].mean())
    bin_size.append(len(idx))

fig, ax = plt.subplots(figsize=(7, 7))

# Perfect calibration reference
ax.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.5,
    label="Perfect calibration"
)

# Observed reliability
ax.plot(
    bin_confidence,
    bin_accuracy,
    marker="o",
    linewidth=2,
    label="AA-EvidentNet (evidential)"
)

for x, y, n in zip(
    bin_confidence,
    bin_accuracy,
    bin_size
):
    ax.annotate(
        f"n={n}",
        (x, y),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=7
    )

ax.set_xlabel("Mean Evidential Confidence")
ax.set_ylabel("Empirical Accuracy")

ax.set_title(
    "AA-EvidentNet Evidential Reliability Diagram"
)

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)

ax.grid(
    linestyle="--",
    alpha=0.3
)

ax.legend(frameon=False)

fig.tight_layout()

out_dir = Path("/content/paper_figures")

png = out_dir / "aa_evidentnet_reliability_final.png"
pdf = out_dir / "aa_evidentnet_reliability_final.pdf"

fig.savefig(png, dpi=300, bbox_inches="tight")
fig.savefig(pdf, bbox_inches="tight")

plt.show()

# Save plotted points so the figure is reproducible
reliability_table = pd.DataFrame({
    "Bin": range(1, 11),
    "Mean Evidential Confidence": bin_confidence,
    "Empirical Accuracy": bin_accuracy,
    "N": bin_size,
})

csv = Path(
    "/content/final_test_backup_upload/"
    "aa_evidentnet_reliability_bins_final.csv"
)

reliability_table.to_csv(csv, index=False)

print(reliability_table.to_string(index=False))
print("\nVisualization: 10 equal-frequency bins")
print("Reported ECE remains: 0.1133 (15 equal-width bins)")
print("Created:", png)
print("Created:", pdf)
print("Saved:", csv)
```
