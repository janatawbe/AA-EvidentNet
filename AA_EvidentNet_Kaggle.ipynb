```python
import os
import pandas as pd
from collections import Counter

repo = "/kaggle/working/AA-EvidentNet"
processed = f"{repo}/data/processed/train"

manifest = pd.read_csv(f"{repo}/data/manifests/train_balanced.csv")
aug = manifest[manifest["is_original"] == False]

actual_by_class = {}

for cls in os.listdir(processed):
    cls_dir = os.path.join(processed, cls)
    if os.path.isdir(cls_dir):
        actual_by_class[cls] = os.listdir(cls_dir)

matched_exact = 0
matched_same_stem = 0
unmatched_examples = []

for rel in aug["path"]:
    cls, expected_name = rel.split("/", 1)
    expected_stem = os.path.splitext(expected_name)[0]

    actual_names = actual_by_class.get(cls, [])

    if expected_name in actual_names:
        matched_exact += 1
        continue

    same_stem = [
        name for name in actual_names
        if os.path.splitext(name)[0] == expected_stem
        or name.startswith(expected_stem)
    ]

    if same_stem:
        matched_same_stem += 1
    elif len(unmatched_examples) < 10:
        unmatched_examples.append((cls, expected_name))

print("Exact matches:", matched_exact)
print("Same-stem / renamed matches:", matched_same_stem)
print("Still unmatched:", len(aug) - matched_exact - matched_same_stem)

print("\nSome unmatched examples:")
for x in unmatched_examples:
    print(x)

print("\nActual filename extensions:")
exts = Counter(
    os.path.splitext(name)[1] or "<no extension>"
    for names in actual_by_class.values()
    for name in names
)
print(exts)
```

```python
import os
import pandas as pd

repo = "/kaggle/working/AA-EvidentNet"
processed = f"{repo}/data/processed/train"

manifest = pd.read_csv(f"{repo}/data/manifests/train_balanced.csv")
aug = manifest[manifest["is_original"] == False]

actual_by_class = {
    cls: os.listdir(os.path.join(processed, cls))
    for cls in os.listdir(processed)
    if os.path.isdir(os.path.join(processed, cls))
}

unique_matches = 0
no_matches = 0
multiple_matches = 0
examples = []

for rel in aug["path"]:
    cls, expected_name = rel.split("/", 1)

    matches = [
        actual_name
        for actual_name in actual_by_class.get(cls, [])
        if expected_name == actual_name
        or expected_name.startswith(actual_name)
        or actual_name.startswith(expected_name)
    ]

    if len(matches) == 1:
        unique_matches += 1
    elif len(matches) == 0:
        no_matches += 1
        if len(examples) < 5:
            examples.append(("NO MATCH", cls, expected_name))
    else:
        multiple_matches += 1
        if len(examples) < 5:
            examples.append(("MULTIPLE", cls, expected_name, matches[:5]))

print("Expected:", len(aug))
print("Unique matches:", unique_matches)
print("No matches:", no_matches)
print("Multiple matches:", multiple_matches)

print("\nExamples of problems:")
for x in examples:
    print(x)
```

```python
import os
import shutil
import pandas as pd
from pathlib import Path

repo = Path("/kaggle/working/AA-EvidentNet")
input_processed = Path(
    "/kaggle/input/datasets/janatawbe/aa-evidentnet-data/prepared/processed"
)
compat = Path("/kaggle/working/processed_compat")

# Start clean
if compat.exists():
    shutil.rmtree(compat)

compat.mkdir(parents=True)

manifest = pd.read_csv(repo / "data/manifests/train_balanced.csv")

expected_processed = [
    Path(p)
    for p in manifest["path"].astype(str)
    if "processed/" in p.replace("\\", "/")
]

created = 0
missing = []
multiple = []

for manifest_path in expected_processed:
    normalized = str(manifest_path).replace("\\", "/")
    relative = Path(normalized.split("processed/", 1)[1])

    expected_target = input_processed / relative
    compat_link = compat / relative
    compat_link.parent.mkdir(parents=True, exist_ok=True)

    # Case 1: Kaggle kept the full filename
    if expected_target.exists():
        os.symlink(expected_target, compat_link)
        created += 1
        continue

    # Case 2: Kaggle truncated the filename
    actual_parent = input_processed / relative.parent

    if not actual_parent.exists():
        missing.append(str(relative))
        continue

    expected_name = relative.name

    candidates = [
        p for p in actual_parent.iterdir()
        if p.is_file() and expected_name.startswith(p.name)
    ]

    if len(candidates) == 1:
        os.symlink(candidates[0], compat_link)
        created += 1
    elif len(candidates) == 0:
        missing.append(str(relative))
    else:
        multiple.append((str(relative), [p.name for p in candidates]))

# Connect repo to compatibility directory
processed_link = repo / "data/processed"

if processed_link.is_symlink() or processed_link.is_file():
    processed_link.unlink()
elif processed_link.is_dir():
    shutil.rmtree(processed_link)

os.symlink(compat, processed_link)

print("Expected augmented files:", len(expected_processed))
print("Compatibility links created:", created)
print("Missing:", len(missing))
print("Multiple matches:", len(multiple))
print("Processed ready:", processed_link.exists())
```

```python
import os
import shutil
import pandas as pd
from pathlib import Path

repo = Path("/kaggle/working/AA-EvidentNet")

input_train = Path(
    "/kaggle/input/datasets/janatawbe/aa-evidentnet-data/prepared/processed/train"
)

compat_train = Path("/kaggle/working/processed_compat/train")

# Start clean
if compat_train.parent.exists():
    shutil.rmtree(compat_train.parent)

compat_train.mkdir(parents=True)

manifest = pd.read_csv(repo / "data/manifests/train_balanced.csv")

# Only augmented rows
augmented = manifest[manifest["is_original"] == False]

created = 0
missing = []
multiple = []

for path_str in augmented["path"].astype(str):

    relative = Path(path_str)

    expected_source = input_train / relative
    link_path = compat_train / relative

    link_path.parent.mkdir(parents=True, exist_ok=True)

    # Kaggle kept full filename
    if expected_source.exists():
        os.symlink(expected_source, link_path)
        created += 1
        continue

    # Kaggle truncated filename
    actual_parent = input_train / relative.parent

    if not actual_parent.exists():
        missing.append(str(relative))
        continue

    expected_name = relative.name

    candidates = [
        p for p in actual_parent.iterdir()
        if p.is_file() and expected_name.startswith(p.name)
    ]

    if len(candidates) == 1:
        os.symlink(candidates[0], link_path)
        created += 1

    elif len(candidates) == 0:
        missing.append(str(relative))

    else:
        multiple.append(
            (str(relative), [p.name for p in candidates])
        )

# Reconnect repo
processed_link = repo / "data/processed"

if processed_link.is_symlink() or processed_link.is_file():
    processed_link.unlink()
elif processed_link.is_dir():
    shutil.rmtree(processed_link)

os.symlink(
    "/kaggle/working/processed_compat",
    processed_link
)

print("Expected augmented files:", len(augmented))
print("Compatibility links created:", created)
print("Missing:", len(missing))
print("Multiple matches:", len(multiple))
print("Processed ready:", processed_link.exists())
```

```python
%cd /kaggle/working/AA-EvidentNet

import os
import time
import shutil
import subprocess
import threading
from pathlib import Path
from datetime import datetime, timezone

repo = Path("/kaggle/working/AA-EvidentNet")
backup = Path("/kaggle/working/aa_live_backup")

console_log = repo / "aa_training_console.log"

# Start training
log_file = open(console_log, "w")

process = subprocess.Popen(
    [
        "python", "run_pipeline.py", "train",
        "--model", "aa_evidentnet",
        "--seed", "42",
        "--device", "cuda",
        "--num-workers", "2",
    ],
    cwd=repo,
    stdout=log_file,
    stderr=subprocess.STDOUT,
)

print("Training PID:", process.pid)
print("Automatic backup: every 30 minutes")


def backup_results():
    while process.poll() is None:
        time.sleep(30 * 60)

        if process.poll() is not None:
            break

        print("\nBacking up checkpoints...")

        checkpoints = repo / "results/checkpoints"
        logs = repo / "results/logs"

        if checkpoints.exists():
            shutil.copytree(
                checkpoints,
                backup / "checkpoints",
                dirs_exist_ok=True
            )

        if logs.exists():
            shutil.copytree(
                logs,
                backup / "logs",
                dirs_exist_ok=True
            )

        if console_log.exists():
            shutil.copy2(console_log, backup / "training_console.log")

        timestamp = datetime.now(timezone.utc).strftime("%Y-%m-%d_%H-%M")

        subprocess.run(
            [
                "kaggle", "datasets", "version",
                "-p", str(backup),
                "-m", f"AA-EvidentNet automatic backup {timestamp}"
            ]
        )

        print("Persistent backup finished.")


threading.Thread(target=backup_results, daemon=True).start()

print("Training started.")
```

```python
from pathlib import Path
import shutil

run_id = "20260831_064112_aa_evidentnet_seed42_48c214"

src_ckpt = Path(f"/kaggle/working/AA-EvidentNet/results/checkpoints/{run_id}")
src_logs = Path(f"/kaggle/working/AA-EvidentNet/results/logs/{run_id}")

stage = Path("/kaggle/working/aa_evidentnet_final_backup")
shutil.rmtree(stage, ignore_errors=True)

(stage / "checkpoints").mkdir(parents=True)
(stage / "logs").mkdir(parents=True)

shutil.copytree(src_ckpt, stage / "checkpoints" / run_id)

if src_logs.exists():
    shutil.copytree(src_logs, stage / "logs" / run_id)

print("Backup staging folder created:")
print(stage)

print("\nCheckpoint files:")
for p in (stage / "checkpoints" / run_id).iterdir():
    print(p.name, round(p.stat().st_size / 1024**2, 2), "MB")
```
